```python
# EXPERIMENT 4 - AIML LAB
# Import, Export, and Dataset Analysis using Pandas
# Step 1: Import Libraries
import pandas as pd
import numpy as np
# Step 2: Load Dataset (Using built-in dataset from seaborn for simplicity)
import seaborn as sns
df = sns.load_dataset('titanic')
# Step 3: Display Dataset
print("----- First 5 Rows -----")
print(df.head())
```

```python
# Step 4: Number of Rows and Columns
print("\n----- Dataset Shape -----")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
# Step 5: Size of Dataset
print("\n----- Dataset Size -----")
print(df.size)
```

```python
# Step 6: Missing Values
print("\n----- Missing Values -----")
print(df.isnull().sum())
# Step 7: Numerical Columns Summary
numeric_df = df.select_dtypes(include=[np.number])
```

```python
# Step 8: Sum
print("\n----- Sum of Numerical Columns -----")
print(numeric_df.sum())

# Step 9: Average (Mean)
print("\n----- Mean of Numerical Columns -----")
print(numeric_df.mean())
```

```python
# Step 10: Minimum Values
print("\n----- Minimum Values -----")
print(numeric_df.min())
# Step 11: Maximum Values
print("\n----- Maximum Values -----")
print(numeric_df.max())
# Step 12: Export Dataset
df.to_csv("exported_titanic.csv", index=False)
print("\nDataset successfully exported as 'exported_titanic.csv'")
```
