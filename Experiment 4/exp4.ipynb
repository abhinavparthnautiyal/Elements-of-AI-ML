```python
import pandas as pd

# Import data from local CSV file
df = pd.read_csv("insurance.csv")

print("Dataset loaded successfully!")
print("\nFirst 5 rows:")
df.head()
```

```python
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])
print("Size (rows×cols)  :", df.size)
print("\nColumn Names:", list(df.columns))
```

```python
print("Missing Values per Column:")
print(df.isnull().sum())
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
numerical_cols = df.select_dtypes(include='number')

print("Numerical Columns:", list(numerical_cols.columns))

print("\nSum:")
print(numerical_cols.sum())

print("\nAverage (Mean):")
print(numerical_cols.mean())

print("\nMinimum Values:")
print(numerical_cols.min())

print("\nMaximum Values:")
print(numerical_cols.max())
```

```python
print("Dataset Info:")
print(df.info())

print("\nStatistical Description of Numerical Columns:")
df.describe()
```

```python
df.to_csv("insurance_exported.csv", index=False)
print("Data exported successfully as 'insurance_exported.csv'")
```
